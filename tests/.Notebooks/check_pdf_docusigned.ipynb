```python
is_docusigned_qry = b"/Reason (Digitally verifiable PDF exported from www.docusign.com)"
```

```python
def is_docusigned(filepath : str) -> bool:
    with open(filepath, "rb") as f:
        return is_docusigned_qry in f.read()
```
